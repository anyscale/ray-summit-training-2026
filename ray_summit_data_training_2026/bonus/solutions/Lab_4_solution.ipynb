## The "hello world" you'll convert
Below is a single-record example using a HuggingFace image-text-to-text
pipeline. This is the kind of code a teammate might paste from a model card.
It is **not** Ray Data — it loads the model every call, processes one row,
and uses no resource hints.

```python
from transformers import pipeline
from PIL import Image

MODEL_NAME = '/mnt/cluster_storage/hf_cache/models--google--gemma-3-4b-it/snapshots/093f9f388b31de276ce2de164bdc2081324b9767'


system_prompt = (
    'You are a helpful assistant. Given a product name and description, and an image of\n'
    'that product, please create a more descriptive and attractive blurb for the product,\n'
    'capturing elements from the image and suitable for use in an ecommerce website.\n'
    'Output a single suggestion only, with no extra conversational language.'
)

def hello_world_caption(item_id: str, desc: str) -> str:
    pipe = pipeline('image-text-to-text', model=MODEL_NAME, device='cuda:0')
    image_path = os.path.join(IMAGE_DIR, f'{item_id}.png')
    messages = [
        {'role': 'system', 'content': [{'type': 'text', 'text': system_prompt}]},
        {'role': 'user', 'content': [
            {'type': 'image', 'path': image_path},
            {'type': 'text',  'text': desc},
        ]},
    ]
    out = pipe(text=[messages], max_new_tokens=256, batch_size=1)
    return out[0][0]['generated_text'][-1]['content']
```

# Reference solution

Ray data pipeline (Prompt A)

```python
import os
import ray
from ray.data import ActorPoolStrategy
from transformers import pipeline

MODEL_NAME = '/mnt/cluster_storage/hf_cache/models--google--gemma-3-4b-it/snapshots/093f9f388b31de276ce2de164bdc2081324b9767'


system_prompt = (
    'You are a helpful assistant. Given a product name and description, and an image of\n'
    'that product, please create a more descriptive and attractive blurb for the product,\n'
    'capturing elements from the image and suitable for use in an ecommerce website.\n'
    'Output a single suggestion only, with no extra conversational language.'
)


class Captioner:
    def __init__(self):
        # Loaded ONCE per actor replica, not once per call
        self.pipe = pipeline('image-text-to-text', model=MODEL_NAME, device='cuda:0')

    def __call__(self, batch: dict) -> dict:
        item_ids = batch['item_id']
        descs = batch['desc']

        # Build one message list per row in the batch
        all_messages = []
        for item_id, desc in zip(item_ids, descs):
            image_path = os.path.join(IMAGE_DIR, f'{item_id}.png')
            messages = [
                {'role': 'system', 'content': [{'type': 'text', 'text': system_prompt}]},
                {'role': 'user', 'content': [
                    {'type': 'image', 'path': image_path},
                    {'type': 'text',  'text': desc},
                ]},
            ]
            all_messages.append(messages)

        # Single batched forward pass
        outs = self.pipe(text=all_messages, max_new_tokens=256, batch_size=len(all_messages))

        # Extract the assistant's final message content for each row
        cat_descs = [o[0]['generated_text'][-1]['content'] for o in outs]

        return {**batch, 'cat_desc': cat_descs}


# Size the actor pool to all available GPUs in the cluster
num_gpus_available = int(ray.cluster_resources().get('GPU', 1))

ds = ray.data.read_parquet(TINY_DIR)

# Repartition so the actor pool has parallel blocks to chew on.
# Aim for ~2-4 blocks per actor so stragglers don't idle the pool.
ds = ds.repartition(num_gpus_available * 4)

ds = ds.map_batches(
    Captioner,
    batch_size=8,
    compute=ActorPoolStrategy(size=num_gpus_available),
    num_gpus=1,
)

ds.write_parquet(OUTPUT)
```

Sanity check (Prompt B)

```python
import ray

ds = ray.data.read_parquet(OUTPUT)

# Show 3 rows with the original desc and the generated cat_desc
sample = ds.take(3)
for row in sample:
    print(f"id: {row['item_id']}")
    print(f"desc:     {row['desc']}")
    print(f"cat_desc: {row['cat_desc']}")
    print('-' * 80)

# Validation check across the full dataset
def check_batch(batch: dict) -> dict:
    descs = batch['desc']
    cat_descs = batch['cat_desc']
    results = []
    for desc, cat_desc in zip(descs, cat_descs):
        ok = (
            isinstance(cat_desc, str)
            and len(cat_desc) > 0
            and len(cat_desc) >= len(desc)
        )
        results.append(ok)
    return {'ok': results, 'desc': descs, 'cat_desc': cat_descs}

checked = ds.map_batches(check_batch, batch_size=64).take_all()

failures = [r for r in checked if not r['ok']]
assert not failures, (
    f"{len(failures)} rows failed validation. First failure: "
    f"desc={failures[0]['desc']!r}, cat_desc={failures[0]['cat_desc']!r}"
)

print(f"\n✓ All {len(checked)} rows passed: cat_desc is a non-empty string "
      f"and at least as long as desc.")
```
