## Wrap-up
- A *deployment* is a class with logic and config (replicas, resources).
- `bind(...)` produces a *bound deployment* — a recipe Ray Serve can construct.
- `serve.run(...)` actually starts replicas; `serve.shutdown()` stops them.
- Replicas are themselves Ray actors. Inspect them in the Ray Dashboard.
